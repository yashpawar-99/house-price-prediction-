```python
import numpy as np 

def suffle_and_split (data,rathio):
    np.random.seed(42)
    suffle = np.random.permutation(len(data))
    split_data = int(len(data)*rathio)
    train_data = suffle[:split_data]
    test_data = suffle[split_data:]

    return data.iloc[train_data],data.iloc[test_data]

train ,test= suffle_and_split(df,0.2)
```

```python
#create income catgory 
df["income cat"] = pd.cut(
    df["median_income"],
    bins=[0.0, 1.5, 3.0, 4.5, 5.0, np.inf],
    labels=[1,2,3,4,5]
)
```

```python
import matplotlib.pyplot as plt
df["income cat"].value_counts().sort_index().plot.bar(rot=0, grid=True)
plt.title("Income Categories Distribution")
plt.xlabel("Income Category")
plt.ylabel("Number of instance")
plt.show()
```

```python
#train test split 

from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["income cat"]
)
```
